**Ejercicio 1: Capacidad de Silos de Granos**

**Problema:** Se requiere calcular cuántas toneladas de maíz puede almacenar un silo cilíndrico. El usuario ingresa el radio y la altura del silo en metros.

**Datos:** Densidad del maíz ≈0.75 ton/m3 . Volumen V=π×r2×h .

**Prueba:** Radio = 3m, Altura = 10m. Salida ≈212.06 toneladas.

```python
import math

#Entrada de datos
radio = float(input("Ingrese el radio del silo en metros: "))
altura = float(input("Ingrese la altura del silo en metros: "))

#proceso
#calculamos el volumen : V = pi *r^2 * h
volumen = math.pi * (radio ** 2) * altura
# calculamos las toneladas: ton = volumen * Dencidad (0.75)
toneladas = volumen * 0.75

#salida
print(f"la capacidad del silo es: {toneladas:.2f} tonelada")
```

**explicaciones de codigo paso a paso**

**Paso 1:** Importar la librería matemáticaSe utiliza import math para poder usar el valor exacto de $\pi$ ($\approx 3.14159$). Sin esto, el cálculo no sería tan preciso.

**Paso 2:** Solicitar los datos al usuario (Entradas)
Se piden el radio y la altura. Es vital usar float() para que el programa entienda que el usuario puede ingresar números con decimales (por ejemplo, 3.5 metros).

**Paso 3:** Calcular el volumen del cilindro
Se aplica la fórmula geométrica:

$$V = \pi \cdot r^2 \cdot h$$

En Python, el radio al cuadrado se escribe como radio ** 2.

**Paso 4:** Calcular el peso en toneladasUna vez que tienes el volumen (espacio que ocupa), lo multiplicas por la densidad del maíz ($0.75$).

Fórmula: $Toneladas = Volumen \times 0.75$.

**Paso 5:** Mostrar el resultado (Salida)
Se imprime el mensaje final. El uso de :.2f dentro de la cadena de texto sirve para redondear el resultado a solo 2 decimales y que sea más fácil de leer.

**Ejercicio 2: Dosificación de Fertilizante Líquido**

**Problema:** Un sistema de riego por goteo requiere aplicar una dosis de fertilizante por hectárea. El usuario ingresa las hectáreas totales y la dosis requerida (litros/hectárea).

**Datos:** El fertilizante se diluye en agua en una relación 1:500 (1 litro de producto por 500 de agua).

**Prueba:** 5 hectáreas, Dosis 2 L/ha. Salida: 10L de producto y 5000L de agua total.

```python
# Entrada de datos
hectareas = float(input("Ingrese las hectáreas totales: "))
dosis_por_ha = float(input("Ingrese la dosis (L/ha): "))

# Proceso
producto_puro = hectareas * dosis_por_ha
agua_necesaria = producto_puro * 500

# Salida
print(f"Producto necesario: {producto_puro} L")
print(f"Agua necesaria para la dilución: {agua_necesaria} L")
```

**explicacion de codigos paso a paso**



*   hectareas = float(input(...)): Convierte la entrada a float (puedes tener 1.5 hectareas).
* dosis_por_ha = float(input(...)): Convierte la entrada a float.   

*   producto_puro = hectareas * dosis_por_ha:
    * Multiplicación de dos float, el resultado es float.

*   agua_necesaria = producto_puro * 500:
    * 500 es un int (entero), pero al multiplicarlo por un float, el resultado final (agua_necesaria) será float.

**Ejercicio 3: Rendimiento de Pulpa en Frutas**

**Problema:** Una planta procesadora necesita saber el porcentaje de aprovechamiento. Se ingresa el peso total de la fruta comprada y el peso de los residuos (cáscaras y semillas).

**Datos:** Pulpa=Peso Total−Residuos. %Rendimiento=(Pulpa/Peso Total)×100.

**Prueba:** Peso Total = 500kg, Residuos = 150kg. Salida: 350kg de pulpa (70%).

```python
# Entrada de datos
peso_total = float(input("Ingrese el peso total de la fruta (kg): "))
residuos = float(input("Ingrese el peso de los residuos (kg): "))

# Proceso
pulpa_util = peso_total - residuos
rendimiento = (pulpa_util / peso_total) * 100

# Salida
print(f"Peso de la pulpa obtenida: {pulpa_util} kg")
print(f"El porcentaje de rendimiento es: {rendimiento:.1f}%")
```

**explicacion de codigos paso a paso**

* peso_total = float(input(...)): Tipo float.

* residuos = float(input(...)): Tipo float.

* pulpa_util = peso_total - residuos: Resta de float, resultado float.

* rendimiento = (pulpa_util / peso_total) * 100:

  * La división siempre genera un float en Python, incluso si el resultado es exacto.

* print(f"...{rendimiento:.1f}%"):

  * Muestra el resultado como un str con un solo decimal.
